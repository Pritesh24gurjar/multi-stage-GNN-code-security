```python
import os

root_folder = "/home/sxi219/RAI-Account/multi-stage-GNN-code-security/IPAG-GIN-CFEExplainer/data/graphs"

status_0 = {}
status_1 = {}

for cwe_id in os.listdir(root_folder):
    cwe_path = os.path.join(root_folder, cwe_id)
    if not os.path.isdir(cwe_path):
        continue
    # Flags for CWE completion (start as True, set to False if any CVE is incomplete)
    complete_0 = True
    complete_1 = True
    for cve_id in os.listdir(cwe_path):
        cve_path = os.path.join(cwe_path, cve_id)
        if not os.path.isdir(cve_path):
            continue
        for label in ["0", "1"]:
            final_path = os.path.join(cve_path, label)
            if not os.path.isdir(final_path):
                # Consider missing directory as incomplete
                if label == "0":
                    complete_0 = False
                else:
                    complete_1 = False
                continue
            files = os.listdir(final_path)
            num_feats = len([f for f in files if "feature" in f])
            # If 3 files per processed graph (nodes, edges, features)
            expected_chunks = len(files) // 3
            if num_feats != expected_chunks:
                if label == "0":
                    complete_0 = False
                else:
                    complete_1 = False
    status_0[cwe_id] = complete_0
    status_1[cwe_id] = complete_1

for cwe in status_0:
    st_0 = status_0[cwe]
    st_1 = status_1.get(cwe, False)  # Handle if key missing in status_1
    if st_0 and st_1:
        print(f"✅ {cwe} is fully processed")
    else:
        if st_0:
            print(f"✅ {cwe} is fully built for negative samples")
        # else:
        #     print(f"❌ {cwe} is not yet fully built for negative samples")

        if st_1:
            print(f"✅ {cwe} is fully built for positive samples")
        # else:
        #     print(f"❌ {cwe} is not yet fully built for positive samples")

```
