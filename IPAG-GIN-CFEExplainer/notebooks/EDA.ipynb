```python
# File: scripts/analyze_dataset.py

import pickle
import json
from pathlib import Path
from collections import Counter
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

def analyze_dataset(pkl_path, metadata_path=None):
    """
    Comprehensive analysis of the dataset pickle file.
    """
    print("="*80)
    print("DATASET ANALYSIS")
    print("="*80)
    
    # Load data
    print(f"\nLoading dataset from: {pkl_path}")
    with open(pkl_path, 'rb') as f:
        graphs = pickle.load(f)
    
    print(f"✓ Loaded {len(graphs)} graphs")
    
    # Load metadata if available
    idx_to_cwe = None
    if metadata_path and Path(metadata_path).exists():
        print(f"Loading metadata from: {metadata_path}")
        with open(metadata_path, 'rb') as f:
            metadata = pickle.load(f)
        idx_to_cwe = metadata.get('idx_to_cwe', {})
        print(f"✓ Loaded metadata with {len(idx_to_cwe)} CWE classes")
    
    # Extract labels
    print("\nExtracting class labels...")
    cwe_indices = [g['cwe_idx'] for g in graphs]
    cwe_ids = [g['cwe_id'] for g in graphs]
    languages = [g['lang'] for g in graphs]
    
    # Count frequencies
    cwe_idx_counts = Counter(cwe_indices)
    cwe_id_counts = Counter(cwe_ids)
    lang_counts = Counter(languages)
    
    # ============================================================================
    # 1. CLASS FREQUENCY STATISTICS
    # ============================================================================
    print("\n" + "="*80)
    print("CLASS FREQUENCY STATISTICS")
    print("="*80)
    
    counts = list(cwe_idx_counts.values())
    print(f"\nCWE Class Distribution (by index):")
    print(f"  Total unique classes: {len(cwe_idx_counts)}")
    print(f"  Max samples in a class: {max(counts):,}")
    print(f"  Min samples in a class: {min(counts):,}")
    print(f"  Mean samples per class: {np.mean(counts):.1f}")
    print(f"  Median samples per class: {np.median(counts):.1f}")
    print(f"  Std dev: {np.std(counts):.1f}")
    
    # Classes below thresholds
    print(f"\nClass size distribution:")
    thresholds = [1, 2, 5, 10, 20, 50, 100, 500, 1000]
    for thresh in thresholds:
        count = sum(1 for c in counts if c >= thresh)
        pct = count / len(counts) * 100
        print(f"  Classes with >= {thresh:4d} samples: {count:3d} ({pct:5.1f}%)")
    
    # ============================================================================
    # 2. TOP AND BOTTOM CLASSES
    # ============================================================================
    print("\n" + "="*80)
    print("TOP 20 MOST FREQUENT CWE CLASSES")
    print("="*80)
    
    print(f"\n{'Rank':<6}{'CWE ID':<15}{'Index':<8}{'Count':>10}{'Percentage':>12}")
    print("-" * 60)
    
    for rank, (cwe_idx, count) in enumerate(cwe_idx_counts.most_common(20), 1):
        cwe_id = idx_to_cwe.get(cwe_idx, f"Unknown") if idx_to_cwe else f"idx_{cwe_idx}"
        pct = count / len(graphs) * 100
        print(f"{rank:<6}{cwe_id:<15}{cwe_idx:<8}{count:>10,}{pct:>11.2f}%")
    
    print("\n" + "="*80)
    print("BOTTOM 20 LEAST FREQUENT CWE CLASSES")
    print("="*80)
    
    print(f"\n{'Rank':<6}{'CWE ID':<15}{'Index':<8}{'Count':>10}{'Percentage':>12}")
    print("-" * 60)
    
    for rank, (cwe_idx, count) in enumerate(sorted(cwe_idx_counts.items(), key=lambda x: x[1])[:20], 1):
        cwe_id = idx_to_cwe.get(cwe_idx, f"Unknown") if idx_to_cwe else f"idx_{cwe_idx}"
        pct = count / len(graphs) * 100
        print(f"{rank:<6}{cwe_id:<15}{cwe_idx:<8}{count:>10,}{pct:>11.2f}%")
    
    # ============================================================================
    # 3. FILTERABLE CLASSES ANALYSIS
    # ============================================================================
    print("\n" + "="*80)
    print("CLASSES BELOW MINIMUM SAMPLE THRESHOLDS (FOR FILTERING)")
    print("="*80)
    
    min_thresholds = [2, 5, 10, 20, 50]
    
    for min_thresh in min_thresholds:
        below_thresh = [c for c in cwe_idx_counts.values() if c < min_thresh]
        removed_samples = sum(below_thresh)
        kept_classes = len([c for c in cwe_idx_counts.values() if c >= min_thresh])
        kept_samples = sum([c for c in cwe_idx_counts.values() if c >= min_thresh])
        
        print(f"\nIf min_samples_per_class = {min_thresh}:")
        print(f"  Classes removed: {len(below_thresh)}")
        print(f"  Classes kept: {kept_classes}")
        print(f"  Samples removed: {removed_samples:,} ({removed_samples/len(graphs)*100:.2f}%)")
        print(f"  Samples kept: {kept_samples:,} ({kept_samples/len(graphs)*100:.2f}%)")
    
    # ============================================================================
    # 4. LANGUAGE DISTRIBUTION
    # ============================================================================
    print("\n" + "="*80)
    print("LANGUAGE DISTRIBUTION")
    print("="*80)
    
    print(f"\n{'Language':<15}{'Count':>10}{'Percentage':>12}")
    print("-" * 40)
    
    for lang, count in lang_counts.most_common():
        pct = count / len(graphs) * 100
        print(f"{lang:<15}{count:>10,}{pct:>11.2f}%")
    
    # ============================================================================
    # 5. GRAPH COMPLEXITY STATISTICS
    # ============================================================================
    print("\n" + "="*80)
    print("GRAPH COMPLEXITY STATISTICS")
    print("="*80)
    
    node_counts = [len(g['ipag_nodes']) for g in graphs]
    edge_counts = [len(g['ipag_edges']) for g in graphs]
    
    print(f"\nNode Statistics:")
    print(f"  Mean nodes per graph: {np.mean(node_counts):.1f}")
    print(f"  Median nodes per graph: {np.median(node_counts):.1f}")
    print(f"  Min nodes: {min(node_counts)}")
    print(f"  Max nodes: {max(node_counts)}")
    print(f"  Std dev: {np.std(node_counts):.1f}")
    
    print(f"\nEdge Statistics:")
    print(f"  Mean edges per graph: {np.mean(edge_counts):.1f}")
    print(f"  Median edges per graph: {np.median(edge_counts):.1f}")
    print(f"  Min edges: {min(edge_counts)}")
    print(f"  Max edges: {max(edge_counts)}")
    print(f"  Std dev: {np.std(edge_counts):.1f}")
    
    # ============================================================================
    # 6. SAVE DETAILED REPORT
    # ============================================================================
    report = {
        'total_graphs': len(graphs),
        'total_classes': len(cwe_idx_counts),
        'class_frequencies': {str(k): v for k, v in cwe_idx_counts.items()},
        'class_frequencies_by_cwe_id': {str(cwe_id_counts[cwe_ids[i]]): cwe_idx_counts[i] for i in range(len(graphs))},
        'language_distribution': dict(lang_counts),
        'node_stats': {
            'mean': float(np.mean(node_counts)),
            'median': float(np.median(node_counts)),
            'min': int(min(node_counts)),
            'max': int(max(node_counts)),
            'std': float(np.std(node_counts))
        },
        'edge_stats': {
            'mean': float(np.mean(edge_counts)),
            'median': float(np.median(edge_counts)),
            'min': int(min(edge_counts)),
            'max': int(max(edge_counts)),
            'std': float(np.std(edge_counts))
        }
    }
    
    report_path = Path(pkl_path).parent / 'dataset_analysis_report.json'
    with open(report_path, 'w') as f:
        json.dump(report, f, indent=2)
    
    print(f"\n✓ Detailed report saved to: {report_path}")
    
    # ============================================================================
    # 7. GENERATE PLOTS
    # ============================================================================
    print("\nGenerating visualizations...")
    
    plot_dir = Path(pkl_path).parent / 'dataset_analysis_plots'
    plot_dir.mkdir(exist_ok=True)
    
    # Plot 1: Class frequency histogram
    fig, ax = plt.subplots(figsize=(14, 6))
    sorted_counts = sorted(cwe_idx_counts.values(), reverse=True)
    ax.bar(range(len(sorted_counts)), sorted_counts)
    ax.set_xlabel('CWE Class (sorted by frequency)', fontsize=12)
    ax.set_ylabel('Number of Samples', fontsize=12)
    ax.set_title('Class Frequency Distribution (Descending)', fontsize=14, fontweight='bold')
    ax.set_yscale('log')
    ax.grid(True, alpha=0.3, axis='y')
    plt.tight_layout()
    plt.savefig(plot_dir / 'class_frequency_histogram.png', dpi=300, bbox_inches='tight')
    plt.close()
    print(f"  ✓ Saved: class_frequency_histogram.png")
    
    # Plot 2: Top 30 classes
    fig, ax = plt.subplots(figsize=(14, 8))
    top_30 = cwe_idx_counts.most_common(30)
    classes = [idx_to_cwe.get(c[0], f"idx_{c[0]}") if idx_to_cwe else f"idx_{c[0]}" for c in top_30]
    counts = [c[1] for c in top_30]
    ax.barh(range(len(classes)), counts, color='steelblue')
    ax.set_yticks(range(len(classes)))
    ax.set_yticklabels(classes)
    ax.set_xlabel('Number of Samples', fontsize=12)
    ax.set_title('Top 30 Most Frequent CWE Classes', fontsize=14, fontweight='bold')
    ax.invert_yaxis()
    ax.grid(True, alpha=0.3, axis='x')
    plt.tight_layout()
    plt.savefig(plot_dir / 'top_30_classes.png', dpi=300, bbox_inches='tight')
    plt.close()
    print(f"  ✓ Saved: top_30_classes.png")
    
    # Plot 3: Language distribution pie chart
    fig, ax = plt.subplots(figsize=(10, 8))
    langs = list(lang_counts.keys())
    counts_lang = list(lang_counts.values())
    colors = plt.cm.Set3(range(len(langs)))
    ax.pie(counts_lang, labels=langs, autopct='%1.1f%%', colors=colors, startangle=90)
    ax.set_title('Language Distribution', fontsize=14, fontweight='bold')
    plt.tight_layout()
    plt.savefig(plot_dir / 'language_distribution.png', dpi=300, bbox_inches='tight')
    plt.close()
    print(f"  ✓ Saved: language_distribution.png")
    
    # Plot 4: Node/Edge complexity
    fig, axes = plt.subplots(1, 2, figsize=(15, 5))
    
    axes[0].hist(node_counts, bins=50, color='steelblue', edgecolor='black', alpha=0.7)
    axes[0].set_xlabel('Number of Nodes', fontsize=12)
    axes[0].set_ylabel('Frequency', fontsize=12)
    axes[0].set_title('Distribution of Nodes per Graph', fontsize=12, fontweight='bold')
    axes[0].grid(True, alpha=0.3, axis='y')
    
    axes[1].hist(edge_counts, bins=50, color='coral', edgecolor='black', alpha=0.7)
    axes[1].set_xlabel('Number of Edges', fontsize=12)
    axes[1].set_ylabel('Frequency', fontsize=12)
    axes[1].set_title('Distribution of Edges per Graph', fontsize=12, fontweight='bold')
    axes[1].grid(True, alpha=0.3, axis='y')
    
    plt.tight_layout()
    plt.savefig(plot_dir / 'graph_complexity.png', dpi=300, bbox_inches='tight')
    plt.close()
    print(f"  ✓ Saved: graph_complexity.png")
    
    print(f"\n✓ All visualizations saved to: {plot_dir}")
    
    print("\n" + "="*80)
    print("✓ ANALYSIS COMPLETE")
    print("="*80)
    
    return {
        'total_graphs': len(graphs),
        'class_counts': cwe_idx_counts,
        'language_counts': lang_counts,
        'node_counts': node_counts,
        'edge_counts': edge_counts
    }


metadata_path = "/home/sxi219/RAI-Account/multi-stage-GNN-code-security/IPAG-GIN-CFEExplainer/data/splits/dataset_metadata.pkl"
source_path = "/home/sxi219/RAI-Account/multi-stage-GNN-code-security/IPAG-GIN-CFEExplainer/data/splits/processed_graphs_multi_class.pkl"
analyze_dataset(source_path, metadata_path)
```
